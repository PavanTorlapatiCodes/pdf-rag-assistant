```python
import gradio as gr

def rag_chat(message, history):
    try:
        # Invoke RAG chain
        response = qa_chain.invoke({
            "query": message
        })

        # Get answer
        answer = response["result"]

        # Get source documents
        source_documents = response.get("source_documents", [])

        # Prepare source information
        sources = []

        for doc in source_documents:
            source = doc.metadata.get("source", "Unknown")
            page = doc.metadata.get("page", "N/A")

            source_info = f"{source} - Page: {page}"

            if source_info not in sources:
                sources.append(source_info)

        # Add sources to final answer
        if sources:
            answer += "\n\n### Sources\n"

            for source in sources:
                answer += f"- {source}\n"

        return answer

    except Exception as e:
        return f"Error: {str(e)}"

demo = gr.ChatInterface(
    fn=rag_chat,
    title="📚 PDF RAG Assistant",
    description="Ask questions about your PDF document.",
    examples=[
        "Give me a summary of this PDF.",
        "What are the main topics in this document?",
        "Give me summary of all functions from this PDF.",
        "Explain the important concepts."
    ]
)


# --------------------------------------------------
# Launch Application
# --------------------------------------------------

if __name__ == "__main__":
    demo.launch()
```

```python

import os
import gradio as gr
from getpass import getpass

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_groq import ChatGroq
from langchain_community.vectorstores import FAISS
from langchain_classic.chains import RetrievalQA
from langchain_community.embeddings import HuggingFaceEmbeddings


# ============================================================
# 0. OpenAI API Key
# ============================================================

if not os.environ.get("GROQ_API_KEY"):
    os.environ["GROQ_API_KEY"] = getpass("ENTER UR KEY")

if not os.environ["GROQ_API_KEY"].strip():
    raise ValueError("API key is required.")


# ============================================================
# Global QA Chain
# ============================================================

qa_chain = None


# ============================================================
# 1. Upload PDF and Create RAG
# ============================================================

def process_pdf(pdf_file):
    global qa_chain

    if pdf_file is None:
        return "❌ Please upload a PDF file."

    try:
        # Load PDF
        loader = PyPDFLoader(pdf_file)
        documents = loader.load()

        if not documents:
            return "❌ No readable pages found in the PDF."

        # Split PDF
        text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=1000,
            chunk_overlap=200
        )
        chunks = text_splitter.split_documents(documents)

        if not chunks:
            return "❌ No readable text found in the PDF."

        # Embeddings
        embeddings = HuggingFaceEmbeddings()
        # Vector Database
        vectorstore = FAISS.from_documents(
            chunks,
            embeddings
        )

        # Retriever
        retriever = vectorstore.as_retriever(
            search_kwargs={"k": 5}
        )

        # LLM
        llm = ChatGroq(
            model="qwen/qwen3.8-27b",
            temperature=0.1
        )

        # RAG Chain
        qa_chain = RetrievalQA.from_chain_type(
            llm=llm,
            chain_type="stuff",
            retriever=retriever,
            return_source_documents=True
        )

        return f"""
### ✅ PDF Ready!

**Pages:** {len(documents)}
**Chunks:** {len(chunks)}

You can now ask questions about your PDF.
"""

    except Exception as e:
        qa_chain = None
        return f"❌ Error: {str(e)}"


# ============================================================
# 2. Ask Question
# ============================================================

def ask_pdf(question):
    global qa_chain

    if qa_chain is None:
        return "⚠️ Please upload and process a PDF first."

    if not question or not question.strip():
        return "Please enter a question."

    try:
        response = qa_chain.invoke({
            "query": question
        })

        answer = response["result"]

        # Get source pages
        source_documents = response.get(
            "source_documents", []
        )

        pages = sorted({
            doc.metadata["page"] + 1
            for doc in source_documents
            if doc.metadata.get("page") is not None
        })

        if pages:
            answer += "\n\n---\n"
            answer += "**📄 Source Pages:** "
            answer += ", ".join(map(str, pages))

        return answer

    except Exception as e:
        return f"❌ Error: {str(e)}"


# ============================================================
# 3. Gradio UI
# ============================================================

with gr.Blocks(title="PDF RAG Assistant") as demo:

    gr.Markdown("""
    # 📚 PDF RAG Assistant

    **Upload a PDF → Process PDF → Ask Questions**
    """)

    pdf_file = gr.File(
        label="📄 Upload PDF",
        file_types=[".pdf"],
        type="filepath"
    )

    process_button = gr.Button("🚀 Process PDF")
    status = gr.Markdown()

    process_button.click(
        fn=process_pdf,
        inputs=pdf_file,
        outputs=status
    )

    gr.Markdown("---")

    question = gr.Textbox(
        label="💬 Ask a Question",
        placeholder="Example: Give me a summary of this PDF"
    )

    ask_button = gr.Button("🔍 Ask PDF")
    answer = gr.Markdown(label="Answer")

    ask_button.click(
        fn=ask_pdf,
        inputs=question,
        outputs=answer
    )

    question.submit(
        fn=ask_pdf,
        inputs=question,
        outputs=answer
    )


# ============================================================
# 4. Launch
# ============================================================

if __name__ == "__main__":
    demo.launch(share=True)
```
